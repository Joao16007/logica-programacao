# Tabela verdade





```python
# And
print(2>1 and 5<8) # V And V - V
print(2>1 and 5>8) # V and F - F
print(2<1 and 5<8) # F and V - F
print(2<1 and 5>8) # F amd F - F
```

```python
# Or
print(2>1 or 5<8) # V or V - V
print(2>1 or 5>8) # V or F - V
print(2<1 or 5<8) # F or V - V
print(2<1 or 5>8) # F or F - F
```

```python
# Not
not 2>1 # V - F
```

```python
# Not
not 1>2 # F - V
```

# Condicional

```python
# if (se)
if 5 < 10:
  print('Verdadeiro')
```

```python
# if/else
num1 = 50
num2 = 10
if num1 > num2:
  print('O num1 é maior que o num2')
else:
  print('O num2 é maior que o num1')

```

```python
# if/elif/else
num1 = 50
num2 = 50
if num1 > num2:
  print('O num1 é maior que o num2')
elif num1 < num2:
  print('O num2 é maior que o num1')
else:
  print('Ambos são iguais')
```

Com operadores lógicos

```python
num1 = 10
num2 = 5
num3 = 2
if (num1 > num2) and (num1 > num3):
  print('O num1 é o maior número')
else:
  print('O num1 não é o maior número')
```

# Estrutura de repetição

##For

```python
# For Range
for contador in range(10):
  print(contador)
```

```python
# For Range + Break
for contador in range (1000):
  print(contador)
  if contador == 2:
    break
```

```python
lista = ['teste1', 'teste2']
```

```python
# for simples
for val in lista:
  print(val)
```

```python
# for com Enumerate
for i, val in enumerate(lista):
  print(i, val)
```

## For com Operadores lógicos

```python
lista = [2,3,4]
```

```python
for val in lista:
  val += 1
  print(val)
```

```python
for val in lista:
  val *= 2
  print(val)
```

## While

```python
val = 1
while(val < 6):
  val *= 2
  print(val)
```

```python
# While com else
val = 1
while(val < 6):
  val *= 2
else:
  print('O valor de val é:', val)
```
